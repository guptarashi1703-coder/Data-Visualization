```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/sharma-kohli.csv")
df
```

```python
plt.plot(df["index"],df["V Kohli"],label="Virat Kohli")
plt.plot(df["index"],df["RG Sharma"],label="Rohit Sharma")
plt.legend(loc="best",fontsize=12)
```

```python
plt.plot(df["index"],df["V Kohli"],label="Virat Kohli")
plt.plot(df["index"],df["RG Sharma"],label="Rohit Sharma")
plt.legend(loc="upper right",fontsize=12)
```

```python
plt.plot(df["index"],df["V Kohli"],label="Virat Kohli",color="red",marker="o",linestyle="dashed",linewidth=6,markersize=10)
plt.plot(df["index"],df["RG Sharma"],label="Rohit Sharma",color="blue",marker="s",linestyle="-",linewidth=6,markersize=10)
plt.legend(loc="best",fontsize=12)
plt.title("Rohit Sharma Vs Virat Kohli",fontsize=14,color="green")
plt.xlabel("Years",fontsize=10,color="red")
plt.ylabel("Run scored",fontsize=10,color="red")
plt.ylim(100,1000)
plt.grid(True,linestyle=":",color="black",alpha=0.5)
plt.savefig("Rohit-Virat.png")
plt.show()
```
